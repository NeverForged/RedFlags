```python
def fix_data():
    '''
    Kill duplicates
    '''
    df = pd.read_csv('data/processed_data.csv')
    # KILL DUPLICATES
    agg_rules = {
        'id': 'max',
        'pred': 'max',
        'date': 'max',
        **{col: 'mean' for col in flag_cols}
    }

    df = df.groupby('comment_text', as_index=False).agg(agg_rules)
    cols = ['toxic', 'obscene', 'threat', 'insult',
           'identity_hate']
    df[cols] = df[flag_cols].round().astype(int)

    # any flagged
    df['flag'] = df[cols].max(axis=1)
    
    df.sort_values('date', ascending=False)
    
    
    df_clean = df[df['flag']==0]
    dirty = {}
    for col in cols:
        dirty[col] = df[df[col]==1]
    
    ndf = pd.DataFrame()
    ndf = df_clean.head(200000)
    for i, col in enumerate(cols):
        N = 10000
        if i == 0:
            N = 20000
        ndf = pd.concat([ndf, dirty[col].head(N)])
    ndf = ndf[['id', 'comment_text', 'toxic', 'obscene', 'threat', 'insult', 'identity_hate', 'pred', 'date']]
    ndf = ndf.drop_duplicates()
    ndf.to_csv('data/processed_data.csv', index=False)
```

```python
fix_data()
```
